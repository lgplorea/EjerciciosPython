```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

notas = pd.read_csv('notas_estudiantes_completo.csv')
notas.head()
```

Crear un barplot mostrando la nota media por asignatura

```python
nota_media = notas.groupby('asignatura')['nota'].mean()
sns.barplot(data=notas, x='asignatura', y='nota', errorbar=None)
plt.title('Nota Media por Asignatura', fontsize = 12, fontweight = 'bold')
plt.xlabel('Asignatura')
plt.ylabel('Nota')
plt.ylim(0,10)
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.show()
```

Crear un boxplot para ver la dispersión.

```python
sns.boxplot(data=notas, x='asignatura', y='nota')
plt.title('Nota Media por Asignatura', fontsize = 12, fontweight = 'bold')
plt.xlabel('Asignatura')
plt.ylabel('Nota')
plt.ylim(0,10)
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.show()
```

Usar countplot  para ver qué asignatura tiene más registros.

```python
sns.countplot(data=notas, x='asignatura')
plt.title('Número de Registros por Asignatura', fontsize=12, fontweight='bold')
plt.xlabel('Asignatura')
plt.ylabel('Número de Registros')
plt.ylim(0,25)
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.show()
```
