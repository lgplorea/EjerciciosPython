```python
import datetime
def edad():
  nombre = input("¿Cómo te llamas?" )
  fecha_nacimiento_str = input("Introduce tu fecha de nacimiento (aaaa/mm/dd): ")

  try:
    año_nacimiento, mes_nacimiento, dia_nacimiento = map(int, fecha_nacimiento_str.split('/'))
    fecha_nacimiento = datetime.date(año_nacimiento, mes_nacimiento, dia_nacimiento)
    fecha_actual = datetime.date(2026, 6, 3)
    años = fecha_actual.year - fecha_nacimiento.year
    if fecha_actual.month < fecha_nacimiento.month or \
       (fecha_actual.month == fecha_nacimiento.month and fecha_actual.day < fecha_nacimiento.day):
        años -= 1

    print(f"Hola {nombre}, tienes {años} años.")
  except ValueError:
    print("Formato de fecha incorrecto. Por favor, usa aaaa/mm/dd.")
  except Exception as e:
    print(f"Ocurrió un error: {e}")
edad()
```
