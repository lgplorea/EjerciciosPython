Creamos un pequeño programa que descuente el inventario de una hamburguesería.

```python
inventario = 100 #definimos la variable inventario y le damos valor con un int de 100
print(f"El inventario contiene {inventario} hamburguesas") #mostramos lo que hay en el inventario

while inventario > 0:
  if inventario <= 10:
    print("El inventario está casi vacío: CUIDADO")
  try:
    num_hamburguesas = int(input("¿Cuántas hamburguesas quiere el cliente?"))
  except ValueError:
    print("Por favor, introduce un número válido")
    continue
  if num_hamburguesas > inventario:
    print("No hay suficientes hamburguesas. Solo hay" ,inventario, "hamburguesas disponibles")
  else:
    inventario -= num_hamburguesas
    print(f"El cliente ha pedido {num_hamburguesas} hamburguesas. El inventario ahora tiene {inventario} hamburguesas")
print("Inventario agotado. Vuelve mañana!")
```
