## En el vasto mundo de la Inteligencia Artificial, los modelos de regresión son una herramienta fundamental para comprender y predecir relaciones entre variables. Al finalizar este taller, tendrás una ligera idea de cómo se construyen y evaluan modelos de regresión, sentando las bases para explorar técnicas más avanzadas en el emocionante campo de la Inteligencia Artificial.

## A través de este ejemplo, aprenderemos los conceptos básicos detrás de la regresión lineal y cómo implementarla en Python utilizando bibliotecas como Pandas, NumPy, Matplotlib y Scikit-learn.

```python
# Importar las bibliotecas necesarias
import pandas as pd  # Para manipulación de datos
import numpy as np  # Para cálculos numéricos
import matplotlib.pyplot as plt  # Para visualización de datos
from sklearn.linear_model import LinearRegression  # Para modelado de regresión lineal
import warnings  # Para gestionar advertencias

# Ignorar todos los warnings
warnings.filterwarnings("ignore")
```

## En este taller introductorio, Utilizaremos un conjunto de datos que relaciona la experiencia laboral con el salario anual.

```python
# Leer el archivo CSV "Salary.csv" y cargar los datos en un DataFrame
df = pd.read_csv("Salary.csv")

# Mostrar las primeras 6 filas del DataFrame para revisar los datos
df.head(6)
```

## Graficamos los datos para observar cual es la correlaciones entre las variables

```python
# Crear un gráfico de dispersión con los datos de "YearsExperience" en el eje x y "Salary" en el eje y
plt.scatter(df["YearsExperience"], df["Salary"])

# Agregar etiquetas al eje x y al eje y para mayor claridad
plt.xlabel("Años de experiencia")  # Etiqueta del eje x
plt.ylabel("Salario anual")  # Etiqueta del eje y

# Mostrar el gráfico
plt.show()

```

## Luego de verificar una correlación positiva y lineal entre las variables, usamos un modelo de regresión lineal simple para analizar datos y realizar predicciones.

```python
# Crear un objeto de modelo de regresión lineal
modelo = LinearRegression()

# Ajustar el modelo utilizando los datos de "YearsExperience" como características y "Salary" como variable objetivo
modelo.fit(df[["YearsExperience"]], df["Salary"])

# Realizar una predicción del salario anual para 15 años de experiencia
prediccion = modelo.predict([[15]])

# Imprimir la predicción
print(f"Predicción del salario anual para 15 años de experiencia: {prediccion}")
```

## Ahora graficamos nuevamente agregando la linea recta que nos aporta el modelo de predicción y a traves de la cual se proyectan o predicen los nuevos datos.

```python
# Crear un gráfico de dispersión con los datos de "YearsExperience" en el eje x y "Salary" en el eje y
plt.scatter(df["YearsExperience"], df["Salary"])

# Superponer la línea de regresión del modelo en el gráfico
plt.plot(df["YearsExperience"], modelo.predict(df[["YearsExperience"]]), color="red")

# Agregar etiquetas al eje x y al eje y para mayor claridad
plt.xlabel("Años de experiencia")  # Etiqueta del eje x
plt.ylabel("Salario anual")  # Etiqueta del eje y

# Mostrar el gráfico
plt.show()
```

## y ahora, usaremos una métrica de evaluación como el coeficiente de determinación (R^2), que nos permite medir la calidad de nuestras predicciones.

```python
# Imprimir el coeficiente de determinación R^2 del modelo, que indica cuánta varianza en la variable dependiente (salario) puede ser explicada por la variable independiente (años de experiencia)
print("R^2: {}".format(modelo.score(df[["YearsExperience"]], df['Salary'])))
```
