```python
from datetime import datetime
from zoneinfo import ZoneInfo

# 1. Definir el menú (Dinámico)
MENUS = {
    "Mañana temprana (6am a 10am)": ["Café con tostadas", "Yogur con avena", "Fruta picada"],
    "Mañana tardía (10am a 1pm)": [],
    "Mediodía / Almuerzo (1pm a 4pm)": ["Menú del día (Plato fuerte)", "Ensalada completa", "Plato de pasta"],
    "Tarde (4pm a 8pm)": ["Merienda de té con galletas", "Batido de frutas", "Sándwich mixto"],
    "Noche (8pm a 11pm)": ["Cena ligera (Pescado/Pollo)", "Tortilla francesa", "Sushi"],
    "Madrugada (11pm a 6am)": []
}

# 2. Obtener la hora actual de España (por defecto)
hora_espana = datetime.now(ZoneInfo("Europe/Madrid"))

print(f"--- SISTEMA DE MENÚS ---")
print(f"Hora actual en España: {hora_espana.strftime('%H:%M')}")

# Preguntar al usuario si quiere cambiar la hora para pruebas
modo = input("¿Deseas usar la hora actual (A) o introducir una hora manual (M)?: ").strip().upper()

if modo == 'M':
    while True:
        hora_manual_str = input("Introduce la hora en formato HH:MM (ejemplo, 14:30): ").strip()
        try:
            # Validamos el formato de hora introducido
            hora_validada = datetime.strptime(hora_manual_str, "%H:%M")
            # Extraemos solo la hora numérica
            hora_actual = hora_validada.hour
            print(f"-> Simulando hora: {hora_manual_str}")
            break
        except ValueError:
            print("Formato incorrecto. Por favor, usa el formato de 24 horas HH:MM (de 00:00 a 23:59).")
else:
    # Si elige 'A' o cualquier otra cosa, se queda con la hora real de España
    hora_actual = hora_espana.hour
    print(f"-> Usando hora actual: {hora_espana.strftime('%H:%M')}")


# 3. Determinar la franja horaria basada en 'hora_actual'
if 6 <= hora_actual < 10:
    franja_actual = "Mañana temprana (6am a 10am)"
elif 10 <= hora_actual < 13:
    franja_actual = "Mañana tardía (10am a 1pm)"
elif 13 <= hora_actual < 16:
    franja_actual = "Mediodía / Almuerzo (1pm a 4pm)"
elif 16 <= hora_actual < 20:
    franja_actual = "Tarde (4pm a 8pm)"
elif 20 <= hora_actual < 23:
    franja_actual = "Noche (8pm a 11pm)"
else:
    franja_actual = "Madrugada (11pm a 6am)"

print(f"Estás en la franja: {franja_actual}\n")

# 4. Obtener las opciones de la franja actual
opciones = MENUS[franja_actual]

# 5. Controlar si hay comida disponible o no
if not opciones:
    print("No es hora de comer, por favor espera un poco. ⏳")
else:
    print("¿Qué te apetece consumir?")
    for indice, plato in enumerate(opciones, start=1):
        print(f"{indice}. {plato}")

    while True:
        try:
            seleccion = int(input(f"\nSelecciona una opción (1 al {len(opciones)}): "))
            if 1 <= seleccion <= len(opciones):
                comida_elegida = opciones[seleccion - 1]
                print(f"\n¡Buena elección! Has elegido: {comida_elegida}")
                break
            else:
                print(f"Por favor, introduce un número válido (entre 1 y {len(opciones)}).")
        except ValueError:
            print("Entrada no válida. Debes introducir un número entero.")
```
