La pandilla de las manzanas: Imagina que un grupo de amigos quiere saber más sobre su cosecha de manzanas. Cosechan 10 manzanasy quiren analizar su peso, para ver cómo es su producción de este año.
Queremos calcular:
- El peso promedio de las manzanas (media).
- El peso central (mediana).
- El peso más común (moda).
- Desviación estándar = qué tanto varían los pesos.

```python
import numpy as np
import statistics as stats

#Pesos de las manzanas en gramos
manzanas = [150, 180, 200, 170, 150, 190, 200, 210, 150, 180]

#Estadísticas descriptivas
media = np.mean(manzanas) # promedio
mediana = np.median(manzanas) # valor central
moda = stats.mode(manzanas) # valor más común
desviacion = np.std(manzanas) # la variabilidad de los pesos

#Resultados
print(f"Analisis de la cosecha de manzanas")
print(f"Peso promedio: {media} gramos")
print(f"Peso central: {mediana} gramos")
print(f"Peso más común: {moda} gramos")
print(f"Desviación estándar: {desviacion: .2f} gramos")
```
