```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

celsius = np.array([-40, -10, 0, 8, 15, 22, 38], dtype=float)
fahrenheit = np.array([-40, 14, 32, 46, 59, 72, 100], dtype=float)
```

```python
modelo = tf.keras.Sequential([
    tf.keras.layers.Input(shape=[1]),
    tf.keras.layers.Dense(units=1)
])

modelo.compile(
	optimizer=tf.keras.optimizers.Adam(0.1),
	loss='mean_squared_error'
)

print("Comenzando entrenamiento...")
historial = modelo.fit(celsius, fahrenheit, epochs=1000, verbose=False)
print("Modelo entrenado")
```

```python
plt.xlabel("# Epoca")
plt.ylabel("Magnitud de pérdida")
plt.plot(historial.history["loss"])

print("Hagamos una predicción")
resultado = modelo.predict(np.array([100.0]))
print("El resultado es " + str(resultado) + " fahrenheit!")
```

```python
print("Variables  internas del modelo")
print(modelo.layers[0].get_weights())
```

por que no usas simplemente modelo.get_weights?

That's a good question! For your current model, which is a tf.keras.Sequential model with only one Dense layer, both modelo.get_weights() and modelo.layers[0].get_weights() will return the same result. This is because modelo.get_weights() returns the weights of all trainable layers in the model, and in your case, there's only one. modelo.layers[0].get_weights() specifically targets the weights of the first layer.

While both work here, modelo.layers[0].get_weights() is more explicit and can be useful if your model had multiple layers and you wanted to inspect the weights of a specific layer by its index.
