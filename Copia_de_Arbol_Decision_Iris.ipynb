# 🌳 Árbol de Decisión Simple con el Dataset Iris

Este cuaderno muestra un ejemplo práctico y sencillo de cómo utilizar un árbol de decisión en Python usando la biblioteca `scikit-learn`.

---

### 📌 Objetivo
Aprender a clasificar datos usando un árbol de decisión, con el clásico conjunto de datos Iris.

```python
# Paso 1: Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
# Paso 2: Cargar el dataset
iris = load_iris()
X = iris.data  # características
y = iris.target  # etiquetas
```

```python
# Paso 3: Dividir los datos en entrenamiento (70%) y prueba (30%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Paso 4: Crear el modelo de árbol de decisión
modelo = DecisionTreeClassifier()
modelo.fit(X_train, y_train)
```

```python
# Paso 5: Hacer predicciones
y_pred = modelo.predict(X_test)
```

```python
# Paso 6: Evaluar el modelo
precision = accuracy_score(y_test, y_pred)
print(f'Precisión del modelo: {precision:.2f}')
```

```python
# Paso 7: Visualizar el árbol
plt.figure(figsize=(12, 8))
plot_tree(modelo, filled=True, feature_names=iris.feature_names, class_names=iris.target_names)
plt.title("Árbol de Decisión - Clasificación de Iris")
plt.show()
```

## Comparación de Precisión con Diferentes `test_size`

Aquí se muestra cómo la precisión puede variar al cambiar el `test_size` a 0.4, lo que significa que el 40% de los datos se usarán para la prueba y el 60% para el entrenamiento.

```python
# Dividir los datos en entrenamiento (60%) y prueba (40%) con test_size=0.4
X_train_0_4, X_test_0_4, y_train_0_4, y_test_0_4 = train_test_split(X, y, test_size=0.4, random_state=42)

# Crear y entrenar un nuevo modelo con este split
modelo_0_4 = DecisionTreeClassifier()
modelo_0_4.fit(X_train_0_4, y_train_0_4)

# Hacer predicciones
y_pred_0_4 = modelo_0_4.predict(X_test_0_4)

# Evaluar el modelo
precision_0_4 = accuracy_score(y_test_0_4, y_pred_0_4)
print(f'Precisión del modelo con test_size=0.4: {precision_0_4:.2f}')
```

Al comparar los resultados con `test_size=0.3`, `test_size=0.2` (que modificamos anteriormente) y ahora `test_size=0.4`, puedes observar cómo el tamaño del conjunto de prueba puede influir en la métrica de precisión. Un `test_size` más pequeño (como 0.2) deja más datos para el entrenamiento, lo que a menudo puede llevar a un modelo más robusto y una mayor precisión en el conjunto de prueba (si los datos son representativos). Por otro lado, un `test_size` más grande (como 0.4) proporciona una evaluación del modelo en una porción más grande de los datos no vistos, pero entrena el modelo con menos datos, lo que a veces puede resultar en una precisión ligeramente menor.
