```python
# 1. Importar las librerías con sus alias convencionales
import pandas as pd
import matplotlib.pyplot as plt

# 2. Cargar el archivo CSV
df = pd.read_csv('notas_estudiantes.csv')

# 3. Agrupar por asignatura y calcular la media de las notas
media_por_asignatura = df.groupby('asignatura')['nota'].mean().reset_index()

# 4. Mostrar el resultado en pantalla
print("--- Nota Media por Asignatura ---")
print(media_por_asignatura)

# 5. CREAR EL GRÁFICO DE BARRAS BÁSICO
# Pasamos primero el eje X (asignaturas) y luego el eje Y (las notas medias)
plt.bar(media_por_asignatura['asignatura'], media_por_asignatura['nota'])

# 6. Personalización
plt.title('Nota media por asignatura',
          fontsize=12,
          fontweight='bold',
          color='#333333')
plt.xlabel('Asignaturas')    # Pone nombre al eje de abajo
plt.ylabel('Nota promedio') # Pone nombre al eje de la izquierda

# 7. MOSTRAR EL GRÁFICO
plt.show()
```

```python
# 2. Filtrar los datos para quedarnos SOLO con el Estudiante_1
estudiante_filtrado = df[df['estudiante'] == 'Estudiante_1']

# 3. Dibujar el gráfico de líneas (Eje X = asignatura, Eje Y = nota)
plt.plot(estudiante_filtrado['asignatura'], estudiante_filtrado['nota'], marker='o', linestyle='-', color='green')

plt.title('Rendimiento académico del Estudiante 1', fontsize=12, fontweight='bold')
plt.xlabel('Asignatura', fontsize=11)
plt.ylabel('Nota', fontsize=11)

# 4. Ajustar el límite del eje Y para que vaya de 0 a 10 obligatoriamente
plt.ylim(0, 10)

# NUEVO: Añadir las notas escritas flotando sobre los puntos
for i, nota in enumerate(estudiante_filtrado['nota']):
    plt.text(i, nota + 0.3, f"{nota:.1f}", ha='center')

# 5. Añadir una cuadrícula tenue de fondo para leer mejor los puntos
plt.grid(True, linestyle='--', alpha=0.5)

# 4. Mostrar el gráfico en pantalla
plt.show()
```
