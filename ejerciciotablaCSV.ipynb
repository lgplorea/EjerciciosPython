```python
import pandas as pd

# Lee el archivo
df = pd.read_csv('ventas.csv')

# Multiplica las dos columnas y guarda el resultado en una nueva llamada 'Total'
df['Total'] = df['precio'] * df['unidades']

# Verificamos que la nueva columna aparezca al final
df.head(10)
```

```python
# Agrupar por producto y sumar ingresos
# 3. Agrupamos por 'producto', sumamos la columna 'Total' y reiniciamos el índice
ingresos_por_producto = df.groupby('producto')['Total'].sum().reset_index()

# 4. Cambiamos el nombre de la columna para que se entienda mejor el resultado
ingresos_por_producto = ingresos_por_producto.rename(columns={'Total': 'Ingresos_Totales'})

# 5. Ordenamos los productos de mayor a menor ingreso
ingresos_por_producto = ingresos_por_producto.sort_values(by='Ingresos_Totales', ascending=False)

ingresos_por_producto
```

```python
# Convierte la columna fecha a tipo datetime
df['fecha'] = pd.to_datetime(df['fecha'])

# Verificamos el resultado
print(df.head(10))
```

```python
# Agrupa por mes y producto
resumen = (
    df.groupby(['mes', 'producto'])
      .agg({
          'unidades': 'sum',
          'Total': 'sum'
      })
      .reset_index()
)

print(resumen)
```
