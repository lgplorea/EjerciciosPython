```python
# 1. Creamos la lista vacía para guardar los alumnos
lista_alumnos = []

print("REGISTRO DE ALUMNOS")

# 2. Bucle para introducir datos dinámicamente
while True:
    nombre = input("\nIntroduce el nombre del alumno (o escribe 'salir' para terminar): ")
    if nombre.lower() == 'salir':
        break

    # Pedimos la edad y la convertimos a número entero
    edad = int(input(f"Introduce la edad de {nombre}: "))

    # Pedimos el sexo
    sexo = input(f"Introduce el sexo de {nombre} (M para Masculino / F para Femenino): ")
    # Normalizamos la respuesta a mayúsculas para evitar errores
    sexo = sexo.upper()

    # Guardamos los datos en un diccionario y lo añadimos a la lista
    alumno = {"nombre": nombre, "edad": edad, "sexo": sexo}
    lista_alumnos.append(alumno)
    print(f"✅ ¡{nombre} registrado con éxito!")

# --- FINALIZACIÓN Y CLASIFICACIÓN ---
print("\n=================================")
print("     RESULTADOS DEL REGISTRO     ")
print("=================================")

total_alumnos = len(lista_alumnos)
print(f"Total de registros: {total_alumnos}")
print("---------------------------------")

# 3. Clasificación por Sexo
mujeres = 0
hombres = 0

for alumno in lista_alumnos:
    if alumno["sexo"] == "F":
        mujeres = mujeres + 1
    elif alumno["sexo"] == "M":
        hombres = hombres + 1

print(f"Total Alumnas (Femenino): {mujeres}")
print(f"Total Alumnos (Masculino): {hombres}")
print("---------------------------------")

# 4. Clasificación por Edad
menores = 0
mayores = 0

for alumno in lista_alumnos:
    if alumno["edad"] < 18:
        menores = menores + 1
    else:
        mayores = mayores + 1

print(f"Alumnos Menores de edad (<18): {menores}")
print(f"Alumnos Mayores de edad (>=18): {mayores}")
```

```python
busqueda = input("Introduce el nombre del alumno que deseas buscar: ")

# Variable para saber si encontramos al alumno o no
encontrado = False

for alumno in lista_alumnos:
    # Comparamos los nombres en minúsculas para que no importen las mayúsculas
    if alumno["nombre"].lower() == busqueda.lower():
        print("¡Alumno encontrado! Datos:")
        print(f"Nombre: {alumno['nombre']}")
        print(f"Edad:   {alumno['edad']} años")
        print(f"Sexo:   {alumno['sexo']}")
        encontrado = True
        break  # Salimos del bucle porque ya lo encontramos

if not encontrado:
  print(f"❌ No se encontró ningún alumno con el nombre '{busqueda}'.")
```
